```python
import matplotlib.pyplot as plt
import numpy as np

# Your empirical data
n = np.array([1000, 5000, 10000, 20000, 50000])
execution_time = np.array([0.0, 0.010, 0.029, 0.095, 0.566])

# Set up the plot size and style
plt.figure(figsize=(9, 6))
plt.plot(n, execution_time, 'bo-', label='Measured Time (Insertion Sort)', linewidth=2, markersize=8)

# Fit a quadratic curve (a * n^2 + b * n + c) to show O(n^2) behavior
coeffs = np.polyfit(n, execution_time, 2)
quadratic_fit = np.poly1d(coeffs)
n_smooth = np.linspace(min(n), max(n), 200)
plt.plot(n_smooth, quadratic_fit(n_smooth), 'r--', label='Theoretical Quadratic Fit ($O(n^2)$)', linewidth=2, alpha=0.8)

# Add labels, title, and legend
plt.xlabel('Input Size ($n$)', fontsize=12, fontweight='bold')
plt.ylabel('Execution Time (Seconds)', fontsize=12, fontweight='bold')
plt.title('Insertion Sort Empirical Time Complexity', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)

# Enable grid for easier reading
plt.grid(True, linestyle='--', alpha=0.6)

# Save the plot high-resolution for your presentation/report
plt.savefig('insertion_sort_complexity.png', dpi=300, bbox_inches='tight')

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Observed benchmark data
n = [10, 20, 30, 40, 50]
execution_time = [0.000012, 0.000056, 0.000156, 0.000639, 0.000782]

plt.figure(figsize=(8, 5))

# Plot observed execution times
plt.plot(
    n,
    execution_time,
    marker="o",
    linewidth=2,
    color="#1f77b4",
    label="Observed Time",
)

# Optional: Plot theoretical O(n^3) reference curve scaled to the last point
c = execution_time[-1] / (n[-1] ** 3)
n_curve = np.linspace(min(n), max(n), 100)
plt.plot(
    n_curve,
    c * (n_curve**3),
    linestyle="--",
    color="#d62728",
    label="Theoretical $O(n^3)$ Reference",
)

# Chart styling
plt.title("Matrix Multiplication Time Complexity", fontsize=13, pad=12)
plt.xlabel("Matrix Dimension ($n \\times n$)", fontsize=11)
plt.ylabel("Execution Time (seconds)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(frameon=True)
plt.tight_layout()

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Observed benchmark data
n = [4, 6, 8, 16, 32]
execution_time = [0.000125, 0.000112, 0.000240, 0.000386, 0.001471]

plt.figure(figsize=(8, 5))

# Plot observed execution times
plt.plot(
    n,
    execution_time,
    marker="o",
    linewidth=2,
    color="#2ca02c",
    label="Observed Time",
)

# Strassen theoretical complexity reference: O(n^(log2(7))) ≈ O(n^2.807)
exponent = np.log2(7)
c = execution_time[-1] / (n[-1] ** exponent)
n_curve = np.linspace(min(n), max(n), 100)

plt.plot(
    n_curve,
    c * (n_curve**exponent),
    linestyle="--",
    color="#d62728",
    label=r"Theoretical $O(n^{\log_2 7}) \approx O(n^{2.81})$ Reference",
)

# Chart styling
plt.title("Strassen's Matrix Multiplication Time Complexity", fontsize=13, pad=12)
plt.xlabel("Matrix Dimension ($n \\times n$)", fontsize=11)
plt.ylabel("Execution Time (seconds)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(frameon=True)
plt.tight_layout()

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Observed benchmark data
n = [1000, 5000, 10000, 20000, 50000]
execution_time = [0.000062, 0.000946, 0.000905, 0.001737, 0.005280]

plt.figure(figsize=(8, 5))

# Plot observed execution times
plt.plot(
    n,
    execution_time,
    marker="o",
    linewidth=2,
    color="#e6550d",
    label="Observed Time",
)

# Theoretical logarithmic complexity reference: O(log2(n))
c = execution_time[-1] / np.log2(n[-1])
n_curve = np.linspace(min(n), max(n), 200)

plt.plot(
    n_curve,
    c * np.log2(n_curve),
    linestyle="--",
    color="#3182bd",
    label=r"Theoretical $O(\log_2 n)$ Reference",
)

# Chart styling
plt.title("Binary Search Time Complexity", fontsize=13, pad=12)
plt.xlabel("Input Size ($n$)", fontsize=11)
plt.ylabel("Execution Time (seconds)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(frameon=True)
plt.tight_layout()

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Observed benchmark data
n = [1000, 5000, 10000, 20000, 50000]
execution_time = [0.000090, 0.000635, 0.001110, 0.002302, 0.006287]

plt.figure(figsize=(8, 5))

# Plot observed execution times
plt.plot(
    n,
    execution_time,
    marker="o",
    linewidth=2,
    color="#9467bd",
    label="Observed Time",
)

# Theoretical expected complexity reference: O(n log2 n)
c = execution_time[-1] / (n[-1] * np.log2(n[-1]))
n_curve = np.linspace(min(n), max(n), 200)

plt.plot(
    n_curve,
    c * (n_curve * np.log2(n_curve)),
    linestyle="--",
    color="#1f77b4",
    label=r"Theoretical Expected $O(n \log_2 n)$ Reference",
)

# Chart styling
plt.title("Randomized Quick Sort Time Complexity", fontsize=13, pad=12)
plt.xlabel("Input Size ($n$)", fontsize=11)
plt.ylabel("Execution Time (seconds)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(frameon=True)
plt.tight_layout()

# Display the plot
plt.show()
```
